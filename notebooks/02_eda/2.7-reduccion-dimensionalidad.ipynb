# 2.7 — Reducción de dimensionalidad para visualización

**Aspecto del rúbrica:** *Uso de métodos de reducción de características para
visualización (PCA, t-SNE, …)*.

Proyecta las variables numéricas del panel a 2-3 dimensiones para explorar
clusters y estructura global: ¿los AGEBs/periodos se agrupan por nivel de
marginación, estación del año o intensidad de exposición?

## Técnicas

1. **PCA** (`sklearn.decomposition.PCA`): proyección lineal; el scree plot
   muestra cuánta varianza retiene cada componente; los *loadings* revelan qué
   variables dominan cada eje.
2. **t-SNE** (`sklearn.manifold.TSNE`): proyección no lineal; preserva
   vecindades locales — útil para encontrar clusters. Ojo: es estocástico
   (`random_state`) y no preserva distancias globales.
