- ydata-profiling funciona mejor con columnas **numéricas y datetime**; excluir

## Configuración e imports

```python
from pathlib import Path
import sys

ROOT_DIR = Path.cwd().parent.parent  # notebooks/<seccion>/ -> raiz del proyecto
if str(ROOT_DIR) not in sys.path:
    sys.path.append(str(ROOT_DIR))

import pandas as pd
from ydata_profiling import ProfileReport

from sonora_health_climate_eda.config import PROCESSED_DATA_DIR, REPORTS_DIR

panel = pd.read_parquet(PROCESSED_DATA_DIR / "panel_hermosillo.parquet")
panel.shape
```

## Generación del reporte

```python
REPORTS_DIR.mkdir(parents=True, exist_ok=True)

profile = ProfileReport(
    panel,
    title="Auto EDA — Panel Hermosillo (salud-clima-marginación)",
    minimal=True,  # cambiar a False para el reporte completo
    progress_bar=True,
)
profile.to_file(REPORTS_DIR / "auto_eda_report.html")
profile  # muestra el reporte embebido en el notebook
```
